```python
X = [1, 2, 3]
```

```python
U = np.random.randn(1)
W = np.random.randn(1)
V = np.random.randn(1)
```

```python
h = 0
```

```python
import numpy as np 
np.random.seed(42)
```

```python
X = [1,2,3]
```

```python
input_size = 1
hidden_size = 2
output_size = 1 
```

```python
# input to hidden
U = np.random.randn(hidden_size, input_size)
print(U) 
```

```python
# hidden to hidden
W = np.random.randn(hidden_size,hidden_size)
print(W)
```

```python
#Hidden to Output
V = np.random.randn(output_size,hidden_size)
print(V)
```

```python
h = np.zeros((hidden_size,1))
print(h)
```

```python
x = np.array([[X[0]]])
print(x)
```

```python
h = np.tanh(U@x+W@h)
print(h)
```

```python
h = np.zeros((hidden_size,1))

for value in X:
    x = np.array([[value]])

    #hidden state
    h = np.tanh(U@x + W@h)

    #output 
    y=V@h 

    print("----------------")
    print("Input")
    print(x)

    print("\nHidden State")
    print(h)

    print("\nOutput")
    print(y)
```
