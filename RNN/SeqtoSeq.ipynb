```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.models import Model 
from tensorflow.keras.layers import LSTM,Dense,Input 
```

```python
input_text=["hi","hello","how are you"]
target_text=["ok","food","i am fine"]
```

```python
target_text = ["\t"+txt+"\n" for txt in target_text]
```

```python
# charecter tockenisation
input_chars = sorted(set("".join(input_text)))
target_chars = sorted(set("".join(target_text)))

input_char_idx = {char:idx for idx,char in enumerate(input_chars)}
target_char_idx = {char:idx for idx,char in enumerate(target_chars)}

num_encoder_tokens = len(input_chars)
num_decoder_tokens = len(target_chars)

max_encoder_len = max(len(txt) for txt in input_text)
max_decoder_len = max(len(txt) for txt in target_text)
```

```python
encoder_input_data = np.zeros((len(input_text),max_encoder_len,num_encoder_tokens),dtype='float32')
decoder_input_data = np.zeros((len(target_text),max_encoder_len,num_encoder_tokens),dtype="float32")
decoder_output_data = np.zeros((len(target_text),max_decoder_len,num_encoder_tokens),dtype='float32')
```

```python
for i,(input_text,target_text) in enumerate(zip(input_text,target_text)):
    for t,char in enumerate(input_text):
        encoder_input_data[i,t,input_char_idx[char]]=1
    for t,char in enumerate(target_text):
        decoder_input_data[i,t,target_char_idx[char]]=1
        if t>0:
            decoder_output_data[i,t-1,target_char_idx[char]]=1
```
