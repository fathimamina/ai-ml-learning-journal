```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Model 
from tensorflow.keras.layers import Input,LSTM,Dense 
```

```python
input_texts = ['hello','hi','thanks']
target_texts = ['hola','hola','gracias']
```

```python
#charecter mapping
input_chars = sorted(list(set("".join(input_texts))))
target_chars = sorted(list(set(''.join(target_texts))))

input_token_index = {c:i for i,c in enumerate(input_chars)}
target_token_index = {c:i for i,c in enumerate(target_chars)}
```

```python
#convert text to numbers
max_encoder_len = max(len(txt) for txt in input_texts)
max_decorder_len = max(len(txt) for txt in target_texts)

num_encoder_tokens = len(input_chars)
num_decorder_tokens = len(target_chars)
```

```python
#preparing training data
encoder_input_data = np.zeros((len(input_texts),max_encoder_len,num_encoder_tokens))
decoder_input_data = np.zeros((len(input_texts),max_decorder_len,num_decorder_tokens))
decorder_target_data = np.zeros((len(input_texts),max_decorder_len,num_decorder_tokens))
```
