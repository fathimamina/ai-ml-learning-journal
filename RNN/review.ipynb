```python
daily_sales = np.array([100, 105, 110, 120, 125, 130, 140, 145, 150, 160,
               165, 170, 180, 185, 190, 200, 205, 210, 220, 225])
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
x = daily_sales/1000.0
x
```

```python
X = []
y = []
window = 4

for i in range(len(daily_sales)-window):
    X.append(x[i:i+window])
    y.append(x[i+window])
X = np.array(X).reshape(-1,window,1)
y = np.array(y)
```

```python
X
```

```python
y
```

```python
from tensorflow.keras.layers import LSTM,Dense,SimpleRNN
from tensorflow.keras.models import Sequential

model = Sequential([
    LSTM(32,activation='tanh',input_shape=(window,1)),
    Dense(1)
])
```

```python
model.compile(optimizer='adam',loss='mse')
```

```python
model.fit(X,y,epochs=10)
```
