```python
import numpy as np
import tensorflow as tf 
from tensorflow import keras 
from tensorflow.keras import layers
```

```python
I = 1
love = 2
cats = 3
animals =4
```

```python
X = np.array([[1,2,3]])
y = np.array([[1,2,4]])
```

```python
vocab_size=5
embed_dim=8

inputs = keras.Input(shape=(3,))

#embedding
x = layers.Embedding(vocab_size,embed_dim)(inputs)

#attention
attention = layers.MultiHeadAttention(num_heads=2,key_dim=8)(x,x)
#Add & Norm 
x = layers.LayerNormalization()(x+attention)
#Feed Forward
ffn = layers.Dense(16, activation='relu')(x)
ffn = layers.Dense(8)(ffn)
#output 
outputs = layers.Dense(vocab_size,activation='softmax')(ffn)

model = keras.Model(inputs,outputs)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100,verbose=0)
```

```python
predicted = model.predict(X)
predicted
```

```python
predicted_id = np.argmax(predicted,axis=-1)
print(predicted_id)
```
