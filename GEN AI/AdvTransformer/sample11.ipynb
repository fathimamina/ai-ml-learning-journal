# BERT
### Text Classification
- "I love this movie"       → POSITIVE
- "I hate this movie"       → NEGATIVE

```python
text = [
    "I love this movie",
    "This movie is amazing",
    "I hate this movie",
    "This movie is terrible"
]
labels = [1,1,0,0]
```

```python
import torch
from transformers import AutoTokenizer,AutoModelForSequenceClassification
```

```python
#Load Pretrained BERT
tockenizer = AutoTokenizer.from_pretrained('bert-base-uncased')

model = AutoModelForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=2)
```

```python
inputs = tockenizer(text, padding=True, truncation=True, return_tensors='pt')
```

```python
labels = torch.tensor(labels)
```

```python
#Fine tune
optimizer = torch.optim.AdamW(model.parameters(), lr=2e-5)

for epoch in range(3):
    outputs = model(**inputs, labels=labels)

    loss = outputs.loss 

    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

    print('Loss:',loss.item())
```

```python
test = "I really love this movie!"

test_inputs = tockenizer(
    test,
    return_tensors="pt"
)

outputs = model(**test_inputs)

prediction = torch.argmax(outputs.logits, dim=1)

print(prediction.item())
```

# GPT-style Transformer
- Input:  "English: I love cats"
- Output: "French: J'aime les chats"

```python
texts = [
    "English: I love cats. French: J'aime les chats.",
    "English: I love you. French: Je t'aime.",
    "English: Good morning. French: Bonjour.",
    "English: Thank you. French: Merci."
]
```

```python
from transformers import AutoTokenizer,AutoModelForCausalLM 
```

```python
tokenizer = AutoTokenizer.from_pretrained('gpt2')
model = AutoModelForCausalLM.from_pretrained('gpt2')
```

```python
tokenizer.pad_token = tokenizer.eos_token

inputs = tokenizer(text,padding=True,trucation=True,return_tensors='pt')
```

```python
#Fine Tune
outputs = model(**inputs, labels=inputs['input_ids'])

loss = outputs.loss
loss.backward()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=5e-5)

for epoch in range(3):
    outputs = model(**inputs, labels=inputs['input_ids'])
    loss = outputs.loss
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    print('Loss:',loss.item())
    
```

```python
text = "English: I love cats. French:"
inputs = tockenizer(text, return_tensors='pt')
outputs = model.generate(**inputs,max_new_tokens=10)
print(tokenizer.decode(outputs[0]))
```

# GPT3

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
texts = [
    "English: Hello\nFrench: Bonjour",
    "English: Good morning\nFrench: Bonjour",
    "English: Good night\nFrench: Bonne nuit",
    "English: Thank you\nFrench: Merci",
    "English: Please\nFrench: S'il vous plaît",
    "English: Yes\nFrench: Oui",
    "English: No\nFrench: Non",
    "English: I love you\nFrench: Je t'aime",
    "English: How are you?\nFrench: Comment allez-vous?",
    "English: Goodbye\nFrench: Au revoir"
]
```

```python
tokenizer = AutoTokenizer.from_pretrained('gpt2')
model = AutoModelForCausalLM.from_pretrained('gpt2')
```

```python
tokenizer.pad_token = tokenizer.eos_token
```

```python
inputs = tokenizer(texts,padding=True,truncation=True,return_tensors='pt')
```

```python
optimizer = torch.optim.AdamW(model.parameters(),lr=5e-5)

for epoch in range(20):

    outputs = model(
        input_ids=inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        labels=inputs["input_ids"]
    )

    loss = outputs.loss

    loss.backward()

    optimizer.step()

    optimizer.zero_grad()

    print("Loss:", loss.item())
```

```python
text = "English: Thank you\nFrench:"

inputs_test = tokenizer(
    text,
    return_tensors="pt"
)

output = model.generate(
    **inputs_test,
    max_new_tokens=10
)

print(
    tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )
)
```

# T5

```python
import torch
from transformers import AutoTokenizer,AutoModelForSeq2SeqLM
```

```python
texts = [
    "translate English to French: Hello",
    "translate English to French: Thank you",
    "translate English to French: Good morning",
    "translate English to French: I love you"
]

targets = [
    "Bonjour",
    "Merci",
    "Bonjour",
    "Je t'aime"
]
```

```python
tokenizer = AutoTokenizer.from_pretrained('t5-small')
model = AutoModelForSeq2SeqLM.from_pretrained('t5-small')
```

```python
inputs = tokenizer(texts, padding=True,truncation=True,return_tensors='pt')
labels = tokenizer(targets, padding=True,truncation=True,return_tensors='pt').input_ids
```

```python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=5e-5
)

for epoch in range(10):

    outputs = model(
        **inputs,
        labels=labels
    )

    loss = outputs.loss

    loss.backward()

    optimizer.step()
    optimizer.zero_grad()

    print("Loss:", loss.item())
```

```python
test = "translate English to French: Thank you"
test_input = tokenizer(test,return_tensors='pt')
outputs = model.generate(**test_input, max_new_tokens=10)

print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
