```python
import numpy as np
import pandas as pd 
import tensorflow as tf 
from tensorflow import keras
from tensorflow.keras import layers 
```

```python
#Create token IDs
I=1
love=2
cats=3
animals=4

X = np.array([[I,love,cats]])
y = np.array([[I,love,animals]])
```

```python
############################Build Transformer###################################
vocab_size = 5
embed_dim = 8 

inputs = keras.Input(shape=(3,))

#Embedding 
x = layers.Embedding(vocab_size, embed_dim)(inputs)

#self-attention 
attention = layers.MultiHeadAttention(
    num_heads=2,
    key_dim=8
)(x,x) 

#Add & Norm 
x = layers.LayerNormalization()(x + attention) 

#Feed Forward 
ffn = layers.Dense(16, activation='relu')(x)
ffn = layers.Dense(8)(ffn)

#Add & Norm
x = layers.LayerNormalization()(x + ffn) 

#outputs
outputs = layers.Dense(vocab_size, activation='softmax')(x)

model = keras.Model(inputs,outputs)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100,verbose=0)
```

```python
preidction = model.predict(X)
print(preidction.shape)
```

```python
preidcted_ids = np.argmax(preidction, axis=-1)
print(preidcted_ids)
```

```python
id_to_word = {
    1: "I",
    2: "love",
    3: "cats",
    4: "animals"
}
```

```python
predicted_sentence = " ".join(id_to_word[i] for i in preidcted_ids[0])

print(predicted_sentence)
```

```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
I = 0
love = 1
cats = 2
animals = 3

X = np.array([[I,love,cats]])

y = np.array([[I,love,animals]])
```

```python
#Build Transformer
vocab_size = 4
embed_dim = 8 

inputs = keras.Input(shape=(3,))

#Embedding
x = layers.Embedding(vocab_size,embed_dim)(inputs)
#self attention
attention = layers.MultiHeadAttention(num_heads=2,key_dim=8)(x,x)
#Add&Norm
x = layers.LayerNormalization()(x+attention)
#FeedForward
ffn = layers.Dense(16, activation='relu')(x)
ffn = layers.Dense(8)(ffn)
#Add & Norm
x = layers.LayerNormalization()(x+ffn)
#output
# outputs = layers.Dense(5,activation='sigmoid')(x)
outputs = layers.Dense(vocab_size,activation='softmax')(x)

model = keras.Model(inputs,outputs)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X,y,epochs=100,verbose=0)
```

```python
prediction = model.predict(X)
prediction
```

```python
predicted_id = np.argmax(prediction,axis=-1)
print(predicted_id)
```
