```python
# ==========================================
# Real-Time Face Detection using OpenCV
# ==========================================

# Step 1: Import Libraries
import cv2
import matplotlib.pyplot as plt

# Step 2: Load Haar Cascade Classifier
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

if face_cascade.empty():
    print("Error loading Haar Cascade!")
else:
    print("Haar Cascade Loaded Successfully")

# Step 3: Read Image
image_path = "sample.jpg"  # Replace with your image path

img = cv2.imread(image_path)

if img is None:
    print("Image not found. Check file path.")
else:

    # Step 4: Convert BGR to RGB for Matplotlib
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Display Original Image
    plt.figure(figsize=(8,6))
    plt.imshow(img_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()

    # Step 5: Convert to Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(8,6))
    plt.imshow(gray, cmap="gray")
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.show()

    # Step 6: Detect Faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(30, 30)
    )

    print(f"Faces Detected: {len(faces)}")

    # Step 7: Draw Bounding Boxes
    for (x, y, w, h) in faces:
        cv2.rectangle(
            img_rgb,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            3
        )

    # Step 8: Display Result
    plt.figure(figsize=(8,6))
    plt.imshow(img_rgb)
    plt.title("Detected Faces")
    plt.axis("off")
    plt.show()

    # Step 9: Save Output Image
    output = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)

    cv2.imwrite("face_detection_output.jpg", output)

    print("Output image saved as 'face_detection_output.jpg'")

# ==========================================
# End of Project
# ==========================================
```
