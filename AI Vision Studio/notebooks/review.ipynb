```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
image = cv2.imread('cat.jpg')

img = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
image.shape
```

```python
resizing = cv2.resize(image,(500,200))
plt.imshow(cv2.cvtColor(resizing,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
flip = cv2.flip(image,0)
plt.imshow(cv2.cvtColor(flip,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
cropped = image[200:400,100:200]
plt.imshow(cv2.cvtColor(cropped,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
normalization = cv2.normalize(image, 0,1, cv2.NORM_MINMAX,dtype=np.uint8)

print(normalization.min())
print(normalization.max())
```

```python
blur = cv2.blur(image,ksize=(5,5))

plt.imshow(cv2.cvtColor(blur,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
rotation = cv2.rotate(image,cv2.ROTATE_90_CLOCKWISE)
plt.imshow(cv2.cvtColor(rotation,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
M = np.float32([
    [1,0,100],
    [0,1,50]
])
h,w = image.shape[:2]
translate = cv2.warpAffine(image,M,(w,h))
plt.imshow(cv2.cvtColor(translate,cv2.COLOR_BGR2RGB))
plt.show()
```

```python
scale = cv2.resize(image,None,fx=2,fy=2)

h,w = image.shape[:2]

zoom_view = image[h//2:h, w//2:w]


plt.imshow(cv2.cvtColor(zoom_view,cv2.COLOR_BGR2RGB))
```
