```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('reference.jpg')
```

```python
print(type(image))
print(image.shape)
```

```python
plt.imshow(image)
plt.show()
```

```python
rgb_image = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)
```

```python
plt.imshow(rgb_image)
plt.axis('off')
plt.show()
```

# Resize

```python
resized = cv2.resize(image,(800,600))

print(resized.shape)
```

```python
plt.imshow(resized)
```

# Cropping

```python
# image[start_row:end_row, start_col:end_col]
cropped = image[200:600, 300:900]
```

```python
plt.imshow(cropped)
```

```python
cropped_rgb = cv2.cvtColor(
    cropped,
    cv2.COLOR_BGR2RGB
)
plt.imshow(cropped_rgb)
plt.axis('off')
plt.show()
```

# Flipping

```python
flipped = cv2.flip(image,-1)
rgb = cv2.cvtColor(flipped,cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
```

# Normalize

```python
normalized = cv2.normalize(
    image,
    None,
    0,
    1,
    cv2.NORM_MINMAX,
    dtype=cv2.CV_32F
)
```

```python
print(normalized.min())
print(normalized.max())
```

# Blur

```python
#Average Blur

blurred = cv2.blur(
    image,
    (5,5)
)

plt.imshow(cv2.cvtColor(blurred,cv2.COLOR_BGR2RGB))
```

```python
#Guassian Blur

guassian =cv2.GaussianBlur(
    image,(5,5),0
)

plt.imshow(cv2.cvtColor(guassian, cv2.COLOR_BGR2RGB))
```

```python
#Median Blur

median = cv2.medianBlur(
    image,5
)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print(gray.shape)
```

```python
ret, thresh = cv2.threshold(
    gray,127,255,cv2.THRESH_BINARY
)
```

```python
plt.imshow(thresh,cmap='gray')
plt.show()
```
