# Morphological operations

Morphological operations are tools that change the shape of objects in a black-and-white image.

```python
import numpy as np
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
_,thresh = cv2.threshold(
    gray,127,255,cv2.THRESH_BINARY
)
```

```python
plt.imshow(thresh,cmap='gray')
```

```python
kernal = np.ones((5,5), np.uint8)
print(kernal)
```

Erosion removes pixels from the boundary first.

```python
#Erosion
eroded = cv2.erode(thresh,kernal,iterations=1)
plt.imshow(eroded,cmap='gray')
```

Dilation

Dilation adds pixels to the edges of white objects, making them larger.

```python
dilated = cv2.dilate(thresh,kernal,iterations=1)
plt.imshow(dilated,cmap='gray')
```

# Opening
Opening removes small white noise from an image while keeping the main object.

```python
opening = cv2.morphologyEx(
    thresh,cv2.MORPH_OPEN,
    kernal
)
plt.imshow(opening,cmap='gray')
```

# Closing

Closing fills small holes or gaps inside an object.

```python
closing = cv2.morphologyEx(
    thresh,cv2.MORPH_CLOSE,kernal
)
plt.imshow(closing,cmap='gray')
```
