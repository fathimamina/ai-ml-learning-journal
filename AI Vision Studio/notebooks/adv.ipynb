```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('cat.jpg')
```

# Adding text on images 

```python
texted = cv2.putText(
    image,
    'Hello OpenCV',
    (50,100),
    cv2.FONT_HERSHEY_COMPLEX,
    1, #font size
    (0,255,0), #color
    2 #thickness

)

texted_rgb = cv2.cvtColor(texted, cv2.COLOR_BGR2RGB)

plt.imshow(texted_rgb)
plt.show()
```

#  Layering images & transparency 


```python
img1 = cv2.imread('cat.jpg')
img2 = cv2.imread('reference.jpg')

img1 = cv2.resize(img1,(500,500))
img2 = cv2.resize(img2,(500,500))

alpha=0.5
blended = cv2.addWeighted(img1,alpha,img2,1-alpha,0)

blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

plt.imshow(blended_rgb)
plt.show()
```

# Shape Matching

```python
img1 = cv2.imread('cat.jpg',0)
img2 = cv2.imread('reference.jpg',0)

_,thresh1 = cv2.threshold(img1,127,255,cv2.THRESH_BINARY)
_,thresh2 = cv2.threshold(img2,127,255,cv2.THRESH_BINARY)

counters1,_ = cv2.findContours(thresh1,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
counters2,_ = cv2.findContours(thresh2,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

cnt1 = counters1[0]
cnt2 = counters2[0]

score = cv2.matchShapes(cnt1,cnt2,cv2.CONTOURS_MATCH_I1,0.0)

print(score)
```

```python
import cv2
import matplotlib.pyplot as plt

img1 = cv2.imread('cat.jpg', 0)
img2 = cv2.imread('images.png', 0)

result = cv2.matchTemplate(img1, img2, cv2.TM_CCOEFF_NORMED)

min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)

top_left = max_loc

h, w = img2.shape

img1_color = cv2.cvtColor(img1, cv2.COLOR_GRAY2BGR)

bottom_right = (top_left[0] + w, top_left[1] + h)

cv2.rectangle(img1_color, top_left, bottom_right, (0, 255, 0), 2)

img1_rgb = cv2.cvtColor(img1_color, cv2.COLOR_BGR2RGB)

plt.imshow(img1_rgb)
plt.show()
```
