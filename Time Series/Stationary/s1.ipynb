```python
import pandas as pd

data = {
    'Date':pd.date_range("2024-01-01", periods=24, freq='ME'),
    "Sales": [
        100, 105, 108, 112,
        118, 123, 129, 134,
        140, 147, 153, 160,
        166, 172, 180, 187,
        194, 201, 209, 216,
        224, 233, 241, 250
    ]
}

df = pd.DataFrame(data)
df.set_index(df['Date'], inplace=True)
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
plt.plot(df['Sales'])
plt.title('Sales')
plt.show()
```

```python
df['Rolling_mean'] = df['Sales'].rolling(window=3).mean()
df
```

```python
df['Rolling_Std'] = df['Sales'].rolling(window=3).std()

df
```

```python
plt.figure(figsize=(12,5))

plt.plot(df['Sales'],label='Original')
plt.plot(df['Rolling_mean'], label='Rolling Mean')
plt.plot(df['Rolling_Std'], label="Rolling Std")

plt.legend()
plt.show()
```

- Does the rolling mean remain almost constant? No

- Does the rolling standard deviation remain almost constant? Yes

- Is there an obvious trend? Yes

So, Not Stationary
