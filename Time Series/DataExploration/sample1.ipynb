```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
data = {
    "Date": pd.date_range("2025-01-01", periods=12, freq="ME"),
    "Sales":[120,135,150,145,160,175,190,185,200,210,225,240]
}

df = pd.DataFrame(data)
```

```python
print(df.head())

print(df.info())

print(df.describe())
```

```python
plt.figure(figsize=(8,4))

plt.plot(df["Date"], df["Sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,4))

sns.lineplot(x="Date", y="Sales", data=df, marker="o")

plt.title("Monthly Sales")

plt.grid(True)

plt.show()
```

```python
df.set_index("Date", inplace=True)

result = seasonal_decompose(df["Sales"],
                            model="additive",
                            period=4)

result.plot()

plt.show()
```
