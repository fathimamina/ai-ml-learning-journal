```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("Images/reference.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis("off")
```

```python
h,w = image.shape[:2]

M = cv2.getRotationMatrix2D(
    (w//2, h//2),
    45,
    1
)

rotated =cv2.warpAffine(
    image,
    M,
    (w,h)
)
plt.imshow(rotated)
plt.axis('off')
```
