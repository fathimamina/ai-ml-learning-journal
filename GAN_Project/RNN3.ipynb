```python
import tensorflow as tf 
from tensorflow  import keras 
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
text = """
i love mechine learning
mechine learning is a branch of Artificial intelligence
mechine learning is fantastic
deep learning is a type of mechine learning
"""
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index)+1

print(total_words)
print(tokenizer.word_index)
```

```python
input_sequences = []
for line in text.split('\n'):
    text_seq = tokenizer.texts_to_sequences([line])[0]

    for i in range(1,len(text_seq)):
        sequence = text_seq[:i+1]
        input_sequences.append(sequence)
print(input_sequences)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences 

max_seqence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen = max_seqence_len,
    padding = 'pre'
)
print(input_sequences)
```

```python
X = input_sequences[:,:-1]
y = input_sequences[:,-1]
```

```python
from tensorflow.keras.utils import to_categorical

y = to_categorical(y, num_classes=total_words)
```

```python
y
```

```python
model = Sequential()
model.add(Embedding(total_words,10,input_length=max_seqence_len-1))
model.add(SimpleRNN(100))
model.add(Dense(total_words,activation='sigmoid'))
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(X,y,epochs=200,verbose=1)
```

```python
import numpy as np

seed_text = 'mechine learning'
next_word = 6

for _ in range(next_word):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list],maxlen=max_seqence_len-1, padding='pre')
    predicted = np.argmax(model.predict(token_list,verbose=0), axis=-1)[0]

    output_word = ""
    for word,index in tokenizer.word_index.items():
        if index == predicted:
            output_word = word
            break
    seed_text += " "+output_word
print(seed_text)
```
