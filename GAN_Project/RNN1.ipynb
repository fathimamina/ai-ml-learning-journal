```python
text = """
i love mechine learning
mechine learning is a branch if Artificial intelligence
mechine learning is fantastic
deep learning is a type of mechine learning
"""
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

## 1. Tokenization

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1

print('Total words:',total_words)
print(tokenizer.word_index)
```

## 2. Create Input Sequence

```python
input_sequence = []
for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        sequence = token_list[:i+1]
        input_sequence.append(sequence)
print(input_sequence)
    
```

## 3. Pad Sequence

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

max_sequence_len = max(len(seq) for seq in input_sequence)

input_sequence = pad_sequences(
    input_sequence,
    maxlen = max_sequence_len,
    padding = 'pre'
)
print(input_sequence)
```

## 4. Saperate X & y

```python
X = input_sequence[:,:-1]
y = input_sequence[:,-1]
print('X',X)
print('y',y)
```

## 5. One-Hot Encoding

```python
from tensorflow.keras.utils import to_categorical

y = to_categorical(y, num_classes = total_words)
y
```

## 6. Build RNN

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Embedding, Dense 

model = Sequential()

model.add(Embedding(total_words, 10, input_length=max_sequence_len-1))

model.add(SimpleRNN(100))

model.add(Dense(total_words, activation='softmax'))

```

```python
model.compile(
    optimizer = 'adam',
    loss = 'categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
model.fit(X,y,epochs=200,verbose=1)
```

```python
import numpy as np
#Generate text
seed_text = 'mechine learning'
next_word = 6

for _ in range(next_word):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences(
        [token_list],
        maxlen=max_sequence_len-1,
        padding='pre'
    )
    predicted = np.argmax(
        model.predict(token_list,verbose=0),
        axis=-1
    )[0]

    output_word = ""
    
    for word, index in tokenizer.word_index.items():
        if index == predicted:
            output_word = word 
            break 
    seed_text += " " + output_word
print(seed_text)
```
