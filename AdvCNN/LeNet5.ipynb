# LeNet-5

## What is LeNet-5?

- LeNet-5 is one of the earliest and most influential **Convolutional Neural Networks (CNNs)**.
- Developed by Yann LeCun and his team in 1998.
- Designed primarily for **handwritten digit recognition**.
- Introduced the Conv → Pool → Fully Connected architecture used in modern CNNs.
- Considered a foundational model in Deep Learning and Computer Vision.

---

## Architecture

```text
Input (32×32×1)
    ↓
C1: Conv Layer (6 filters, 5×5)
    ↓
S2: Average Pooling
    ↓
C3: Conv Layer (16 filters, 5×5)
    ↓
S4: Average Pooling
    ↓
Flatten
    ↓
C5: Dense (120)
    ↓
F6: Dense (84)
    ↓
Output Layer (10 classes)
```

---

## Key Features

- Uses Convolution Layers for feature extraction.
- Uses Average Pooling for dimensionality reduction.
- Uses `tanh` activation in the original architecture.
- Designed for grayscale images.
- Lightweight architecture with relatively few parameters.
- Effective for simple image classification tasks.

---

## Advantages

- Simple and easy to understand.
- Fast training and inference.
- Requires less computational power.
- Good for learning CNN fundamentals.
- Performs well on small datasets.
- Easy to implement using TensorFlow or PyTorch.

---

## Disadvantages

- Not suitable for complex image datasets.
- Limited number of layers compared to modern CNNs.
- Uses Average Pooling, which is generally less effective than Max Pooling.
- Lower accuracy on large-scale image recognition tasks.
- Cannot learn highly complex visual features.

---

## Applications

- Handwritten digit recognition (MNIST).
- Character recognition systems.
- Bank cheque processing.
- Postal code recognition.
- Optical Character Recognition (OCR).
- Educational projects for learning CNNs.

---

## Why is LeNet-5 Important?

- One of the first successful CNN architectures.
- Demonstrated automatic feature extraction from images.
- Introduced concepts that are still used in modern CNNs.
- Served as the foundation for architectures such as AlexNet, VGG, and ResNet.

---

## Comparison with Modern CNNs

| Feature | LeNet-5 | Modern CNNs |
|----------|----------|-------------|
| Year | 1998 | 2012+ |
| Depth | Shallow | Deep |
| Pooling | Average Pooling | Mostly Max Pooling |
| Activation | tanh | ReLU |
| Dataset Size | Small | Very Large |
| Accuracy | Moderate | High |
| Computation | Low | High |

---

## Conclusion

- LeNet-5 is the foundation of modern CNN architectures.
- It introduced convolution and pooling concepts that remain important today.
- Although modern CNNs achieve higher accuracy, LeNet-5 is still widely used for educational purposes and understanding CNN fundamentals.

````python
import tensorflow as tf
from tensorflow.keras import layers,models 
from tensorflow.keras.datasets import mnist
````

````python
#Load Dataset
(X_train,y_train), (X_test,y_test) = mnist.load_data()
````

````python
# Normalize
X_train = X_train.astype('float32')/255.0
X_test = X_test.astype('float32')/255.0
````

````python
#Add channel dimension
X_train = X_train[...,tf.newaxis]
X_test = X_test[...,tf.newaxis]
````

````python
#pad 28x28 to 32x32
X_train = tf.pad(X_train, [[0,0],[2,2],[2,2],[0,0]])
X_test = tf.pad(X_test, [[0,0],[2,2],[2,2],[0,0]])

print(X_train.shape)
````

````python
model = models.Sequential([

    #c1
    layers.Conv2D(
        filters=6,
        kernel_size=(5,5),
        activation='tanh',
        input_shape=(32,32,1)
    ),
    #s2
    layers.AveragePooling2D(pool_size=(2,2)),

    #c3
    layers.Conv2D(
        filters=16,
        kernel_size=(5,5),
        activation='tanh',
    ),
    layers.AveragePooling2D(pool_size=(2,2)),
    layers.Flatten(),

    #s4
    layers.Dense(120,activation='tanh'),

    layers.Dense(84,activation='tanh'),

    layers.Dense(10,activation='softmax'),

])
model.summary()
````

````python
model.compile(
    optimizer = 'adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

````python
history = model.fit(X_train,
                    y_train,
                    epochs=5,
                    batch_size=64,
                    validation_split=0.1
                )
````

````python
loss,accuracy = model.evaluate(X_test,y_test)
print("Accuracy:",accuracy)
````
