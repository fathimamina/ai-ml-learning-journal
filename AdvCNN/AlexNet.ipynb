# AlexNet

## What is AlexNet?

- AlexNet is a deep Convolutional Neural Network (CNN) introduced in 2012.
- Developed by Alex Krizhevsky, Ilya Sutskever, and Geoffrey Hinton.
- Won the ImageNet Large Scale Visual Recognition Challenge (ILSVRC) 2012.
- Marked the beginning of the modern Deep Learning revolution in Computer Vision.
- Designed for large-scale image classification tasks.

---

## Why is AlexNet Important?

- Demonstrated the power of Deep Learning for image recognition.
- Significantly outperformed traditional computer vision techniques.
- Popularized the use of GPUs for training deep neural networks.
- Introduced techniques that are still widely used today.

---

## Architecture

```text
Input (227×227×3)
        ↓
Conv1 (96 filters, 11×11)
        ↓
ReLU
        ↓
Max Pooling
        ↓
Conv2 (256 filters, 5×5)
        ↓
ReLU
        ↓
Max Pooling
        ↓
Conv3 (384 filters, 3×3)
        ↓
ReLU
        ↓
Conv4 (384 filters, 3×3)
        ↓
ReLU
        ↓
Conv5 (256 filters, 3×3)
        ↓
ReLU
        ↓
Max Pooling
        ↓
Flatten
        ↓
FC Layer (4096)
        ↓
Dropout
        ↓
FC Layer (4096)
        ↓
Dropout
        ↓
Output Layer (1000 Classes)
```

---

## Key Features

- Uses ReLU activation instead of tanh.
- Uses Max Pooling instead of Average Pooling.
- Contains 5 Convolutional Layers.
- Contains 3 Fully Connected Layers.
- Uses Dropout to reduce overfitting.
- Trained using GPUs for faster computation.
- Designed for RGB images.

---

## New Concepts Introduced

### ReLU (Rectified Linear Unit)

```text
f(x) = max(0, x)
```

Benefits:
- Faster training.
- Reduces vanishing gradient problems.
- Simpler computation.

### Max Pooling

Example:

```text
2 5
1 3
```

Output:

```text
5
```

Benefits:
- Reduces image dimensions.
- Retains the most important features.
- Improves computational efficiency.

### Dropout

- Randomly deactivates neurons during training.
- Helps prevent overfitting.
- Improves model generalization.

---

## Advantages

- Higher accuracy than LeNet-5.
- Can learn more complex image features.
- Faster convergence due to ReLU activation.
- Better performance on large datasets.
- Introduced techniques still used in modern CNNs.

---

## Disadvantages

- Large number of parameters.
- High memory consumption.
- Requires significant computational resources.
- Can overfit if regularization is not used.
- Slower than many modern optimized architectures.

---

## Applications

- Image Classification.
- Object Recognition.
- Face Recognition.
- Medical Image Analysis.
- Autonomous Vehicles.
- Surveillance Systems.
- Industrial Quality Inspection.

## AlexNet vs LeNet-5

| Feature | LeNet-5 | AlexNet |
|----------|----------|----------|
| Year | 1998 | 2012 |
| Conv Layers | 2 | 5 |
| Activation | tanh | ReLU |
| Pooling | Average Pooling | Max Pooling |
| Input Image | 32×32 | 227×227 |
| Dataset Size | Small | Large |
| Parameters | Few | Millions |
| Accuracy | Moderate | High |

---

## Limitations of AlexNet

- Very large model size.
- Computationally expensive.
- Requires more memory than LeNet.
- Has been surpassed by newer architectures such as VGG, ResNet, and EfficientNet.

---

## Conclusion

- AlexNet was a breakthrough CNN architecture that transformed computer vision.
- It introduced ReLU, Max Pooling, Dropout, and GPU-based training.
- It achieved remarkable performance on ImageNet and inspired many modern CNN architectures.
- AlexNet remains an important milestone in the history of Deep Learning.

````python
import tensorflow as tf
from tensorflow.keras import layers,models 
from tensorflow.keras.datasets import mnist 
````

````python
(X_train,y_train),(X_test,y_test) = mnist.load_data()
````

````python
X_train = X_train.astype('float32')/255.0 
X_test = X_test.astype('float32')/255.0
````

````python
X_train = X_train[...,tf.newaxis]
X_test = X_test[...,tf.newaxis]
````

````python
print(X_train.shape)
````

````python
model = models.Sequential([
    layers.Conv2D(
        filters=96,
        kernel_size=(3,3),
        activation='relu',
        input_shape=(28,28,1)
    ),
    layers.MaxPool2D(pool_size=(2,2)),

    layers.Conv2D(
        filters=256,
        kernel_size=(3,3),
        activation='relu',
    ),
    layers.MaxPool2D(pool_size=(2,2)),

    layers.Conv2D(
        filters=384,
        kernel_size=(3,3),
        activation='relu',
        padding='same'
    ),

    layers.Conv2D(
        filters=384,
        kernel_size=(3,3),
        activation='relu',
        padding='same'
    ),

    layers.Conv2D(
        filters=256,
        kernel_size=(3,3),
        activation='relu',
        padding='same'
    ),
    layers.MaxPool2D(pool_size=(2,2)),

    #flatten
    layers.Flatten(),

     # FC1
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),

    # FC2
    layers.Dense(512, activation='relu'),
    layers.Dropout(0.5),

    # Output
    layers.Dense(10, activation='softmax')
    
])
````

````python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

````python
history = model.fit(
    X_train,
    y_train,
    epochs=3,
    batch_size=32,
    validation_split=0.1    
)
````
